# PageRank

## Introducción

No todas las páginas web tienen la misma importancia.

## La importancia de una página web

La importancia de una página web depende del número de páginas que hacen referencia (el número de predecesores) a él.

Cada página tiene un número de sucesores. Si la página A tiene una importancia x con n sucesores, todas las páginas referenciadas tendrán una importancia de referencia x/n.

La noción de predecesoras y sucesoras:
A -> B
A tiene una sucesora B.
B tiene una predecesora A.

Ejemplo:


![page_rank_1](wikipedia_image_page_rank.png)

Entonces, la importancia de una página se define por la suma de las referencias de todos los predecesores.

Ejemplo:

$$A = \frac{B}{2} + \frac{A}{2}$$
$$B = \frac{A}{2} + C$$
$$C = \frac{B}{2}$$

**texto en negrita**## Resolución matricial

La matriz se encuentra escribiendo todas las diferentes relaciones entre todos los diferentes sitios web.
Para calcular nuestra solución, tenemos una matriz estocástica M y un vector r.

La matriz estocástica de tamaño M (n*n) representa todas las relaciones entre todos los sitios web con el encabezado de columna que representa el sitio web y los encabezados de fila los sucesores.

La composición de la matriz sigue las relaciones entre un sitio web y sus sucesores/predecesores.

Composición de la matriz estocástica (M):

|    |    |    |    |    |    |
|----|-----|-----|-----|-----|
|    | A   | B   | C   | D   | E   |
| A  | A->A| B->A   | C->A   | D->A   | E->A|
| B  | A->B| B->B   | C->B   | D->B   | E->B|
| C  | A->C| B->C   | C->C   | D->C   | E->C|
| D  | A->D| B->D   | C->D   | D->D   | E->D|
| E  | A->E| B->E   | C->E   | D->E   | E->E|

Por ejemplo la relación en M[0,3] is D --> A.

la matriz estocástica (M):
$\left[\begin{array}{cccc}
0 & 0 & 0 & 1 & 1/3 \\
1 & 0 & 0 & 0 & 1/3 \\
0 & 1 & 0 & 0 & 0 \\
0 & 0 & 1/2 & 0 & 1/3 \\
0 & 0 & 1/2 & 0 & 0 \\
\end{array} \right]$

El vector (r) tiene el tamaño del número de sitios web. Cada valor en el vector es el mismo 1/n, así que en nuestro caso 1/5 porque tenemos 5 sitios web.

r0 vector:
$\left[ \begin{array}{cccc}
1/5 \\
1/5 \\
1/5 \\
1/5 \\
1/5 \\
\end{array} \right]$

Para obtener nuestra solución con el método matricial, usamos el siguiente algoritmo:
• Inicialización : $r_{0} = [1/N,....,1/N]^{T};$
• Iteración : $r_{k+1} = M \cdot r_{k};$
• Detente cuando |$r_{k+1} - r_{k}|_{L1} < \epsilon ;$

$|x|_{L1}$ es la norma L1 (también podemos usar todas las demás normas vectoriales, como la euclidiana)

Primero, definiremos nuestro algoritmo, para hacerlo creamos una función llamada calcule_page_rank con el parámetro epsilon, M y R0.
La función imprimirá en la consola la salida de cada iteración.

```python
import numpy as np


def calculate_page_rank(M, r0, epsilon):
    num_iteration = 0
    do_loop = True
    rk1 = np.dot(M, r0)
    print(f"iteration r{num_iteration} = " + np.array2string(rk1, precision=2, separator=',', suppress_small=True))
    while do_loop:
        num_iteration += 1
        rk0 = rk1
        rk1 = np.dot(M, rk1)
        print(f"Iteration r{num_iteration} = " + np.array2string(rk1, precision=2, separator=',', suppress_small=True))
        do_loop = not (np.linalg.norm((rk1 - rk0), ord=1) < epsilon)
    return rk1
```

Ahora definimos nuestras matrices y llamamos a la función previamente creada `calculate_page_rank`.

```python
matrix_size = 5

M = np.array(
    [
        [0, 0, 0, 1, 1 / 3],
        [1, 0, 0, 0, 1 / 3],
        [0, 1, 0, 0, 0],
        [0, 0, 1 / 2, 0, 1 / 3],
        [0, 0, 1 / 2, 0, 0]
    ]
)

r0 = np.array(
    [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5]
).transpose()

epsilon = 0.1

pagerank_result = calculate_page_rank(M, r0, epsilon)
```

Los resultados del cálculo en la `Resolución manual` son iguales a la solución del algoritmo en la iteración número 9 (r9). Nuestro algoritmo parece correcto.

$$A = \frac{5}{24} ≈ 0.21$$

$$B = \frac{3}{12} ≈ 0.27$$

$$C = \frac{3}{12} ≈ 0.26$$

$$D = \frac{2}{12} ≈ 0.15$$

$$E = \frac{3}{24} ≈ 0.11$$ 



También podemos mostrar los sitios web más importantes:

```python
dictionary_index_link = {
    '1':'A',
    '2':'B',
    '3':'C',
    '4':'D',
    '5':'E'
}

def display_websites(dictionary_index_link, page_rank_result, matrix_size, row_number=3):
    pagerank = [(dictionary_index_link[str(i)], page_rank_result[i - 1]) for i in range(1, matrix_size + 1)]
    pagerank.sort(key=lambda a: a[1], reverse=True)
    # print(pagerank[:10])
    for i in range(row_number):
        print(f"{i+1} : {pagerank[i]}")

display_websites(dictionary_index_link, pagerank_result, matrix_size, 5)
```

## La trampa de la araña

La trampa de la araña es un problema que ocurre durante la clasificación de la página. Algunos sitios web pueden bloquear a los navegantes web en un grupo de sitios web solo haciendo referencia a sitios web en los mismos grupos que los tienen.
En los ejemplos anteriores, puedes salir y llegar a cualquier sitio web desde cualquier lugar, en el siguiente gráfico es completamente imposible, una vez que llegas a D no puedes salir.

Para resolver este problema, implementamos el teletransporte.

## Teleport

El teletransporte es una forma para que un internauta escape de cualquier trampa de araña. Un navegante de la web saltará aleatoriamente de un sitio web a otro.

Para evitar el problema de la trampa de araña, Google utiliza teletransportes aleatorios.

En cada iteración, el internauta puede:
- Saltar a un enlace aleatorio con una probabilidad $\beta$.
- Saltar uniformemente a una página aleatoria con una probabilidad de $1 - \beta$.

El valor $\beta$ debe establecerse entre 0,8 y 0,9.

Esta solución teletransportará a todos los internautas fuera de cada trampa de araña.

Para implementar el teletransporte en el paso $M \cdot r$, haremos otra operación.

Como antes, tenemos:

Matriz M:
$\left[\begin{array}{cccc}
0 & 0 & 0 & 0 & 1/3 \\
1 & 0 & 0 & 0 & 1/3 \\
0 & 1 & 0 & 0 & 0 \\
0 & 0 & 1/2 & 1 & 1/3 \\
0 & 0 & 1/2 & 0 & 0 \\
\end{array} \right]$

Vector r:
$\left[ \begin{array}{cccc}
1/5 \\
1/5 \\
1/5 \\
1/5 \\
1/5 \\
\end{array} \right]$

Y una nueva matriz que tiene el mismo tamaño tiene M y está llena de 1/5 (1 dividido por el número de sitios web).

Matriz T:
$\left[\begin{array}{cccc}
1/5 & 1/5 & 1/5 & 1/5 & 1/5 \\
1/5 & 1/5 & 1/5 & 1/5 & 1/5 \\
1/5 & 1/5 & 1/5 & 1/5 & 1/5 \\
1/5 & 1/5 & 1/5 & 1/5 & 1/5 \\
1/5 & 1/5 & 1/5 & 1/5 & 1/5 \\
\end{array} \right]$

Esta última matriz es lo que permite a un internauta saltar de una trampa para arañas.
Luego usamos nuestro algoritmo anterior nuevamente con nuestra nueva matriz:
$M' = M*\beta + T*(1-\beta)$

Beta lo define usted y generalmente es 0.8 or 0.9.

```python
def teleport_operation(M, T, beta):
    return M * beta + T * (1 - beta)
```

```python
beta = 0.8
epsilon = 0.1

M = np.array(
    [
        [0, 0, 0, 0, 1 / 3],
        [1, 0, 0, 0, 1 / 3],
        [0, 1, 0, 0, 0],
        [0, 0, 1 / 2, 1, 1 / 3],
        [0, 0, 1 / 2, 0, 0]
    ]
)

T = np.array(
    [
        [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5],
        [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5],
        [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5],
        [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5],
        [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5],
    ]
)

r0 = np.array(
    [1 / 5, 1 / 5, 1 / 5, 1 / 5, 1 / 5]
).transpose()

M = teleport_operation(M, T, beta)

pagerank_result = calculate_page_rank(M, r0, epsilon)
```

```python
display_websites(dictionary_index_link, pagerank_result, matrix_size, 5)
```

Los valores para cada sitio web ahora se distribuyen de manera más uniforme. El sitio web D ya no tiene el 90% del valor de referencia web, ahora tiene el 50%.
